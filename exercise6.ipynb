Tuples


```python
#operations in tuples
t1=(1,2,3,4)
t2=(22,33,44,55,"PRANATHI")
print(t1+t2)
print(t1*3)
print(t2[4])
for i in t1:
  print(i,end=" ")
```

```python
#Functions
t1=(31,4,54,6)
print(len(t1))
print(min(t1))
print(max(t1))
print(tuple(reversed(t1)))
print(tuple(sorted(t1)))
print(sum(t1))
```

```python
#Tuple Methods
t1=(76,90,76,33)
print(t1.count(76))
print(t1.index(90))
```

```python
#zip() function
t1=("n1","n2","n3")
t2=(1,2,3)
print(tuple(zip(t1,t2)))
```
